```python
c1 = np.array([[1, 2], [3, 4]])
c2 = np.array([[1, 2], [3, 4]])
print(np.array_equal(c1, c2))
print(c1 == c2)
```

```python
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
v = np.bincount(x)
m = np.argmax(v)
print(m)
print(np.where(x == m)[0])
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])
v = np.bincount(y)
m = np.argmax(v)
print(m)
print(np.where(y == m)[0])
```

```python
g = np.matrix('[4,1,9;12,3,1;4,5,6]')
print(np.sum(g))
print(np.sum(g, axis=1))
print(np.sum(g, axis=0))
```

```python
n = np.array([[55, 25, 15],
              [30, 44, 2],
              [11, 45, 77]])

print(np.trace(n))
print(np.linalg.eigvals(n))
w, v = np.linalg.eig(n)
print(v)
print(np.linalg.inv(n))
print(np.linalg.det(n))
```

```python
p = np.array([[1, 2],
              [2, 3]])
q = np.array([[4, 5],
              [6, 7]])

print(np.matmul(p, q))
print(np.cov(p, q))
```

```python
p = np.array([[1, 2],
              [2, 3],
              [4, 5]])
q = np.array([[4, 5, 1],
              [6, 7, 2]])

print(np.matmul(p, q))
print(np.cov(p.T, q))
```

```python
x = np.array([[2, 3, 4],
              [3, 2, 9]])

y = np.array([[1, 5, 0],
              [5, 10, 3]])

print(np.inner(x, y))
print(np.outer(x, y))
print(np.transpose([np.repeat(x.flatten(), len(y.flatten())),
                    np.tile(y.flatten(), len(x.flatten()))]))
```

```python
a = np.array([[1, -2, 3],
              [-4, 5, -6]])

print(np.abs(a))

print(np.percentile(a, [25, 50, 75]))
print(np.percentile(a, [25, 50, 75], axis=0))
print(np.percentile(a, [25, 50, 75], axis=1))

print(np.mean(a))
print(np.mean(a, axis=0))
print(np.mean(a, axis=1))

print(np.median(a))
print(np.median(a, axis=0))
print(np.median(a, axis=1))

print(np.std(a))
print(np.std(a, axis=0))
print(np.std(a, axis=1))
```

```python
b = np.array([-1.8, -1.6, -0.5, 0.5, 1.6, 1.8, 3.0])

print(np.floor(b))
print(np.ceil(b))
print(np.trunc(b))
print(np.round(b))
```

```python
from PIL import Image

def img_to_array(path):
    img = Image.open(path)

    if img.mode == "RGB":
        a = np.array(img)
        np.savetxt("rgb_image.txt", a.reshape(-1, 3), fmt="%d")
        print("RGB image saved as rgb_image.txt")

        x = np.loadtxt("rgb_image.txt", dtype=int)
        print(x)

    else:
        img = img.convert("L")
        a = np.array(img)
        np.savetxt("gray_image.txt", a, fmt="%d")
        print("Gray image saved as gray_image.txt")

        x = np.loadtxt("gray_image.txt", dtype=int)
        print(x)

img_to_array("image.jpg")
```
