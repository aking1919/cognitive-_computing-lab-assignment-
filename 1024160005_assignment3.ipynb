```python
import pandas as pd



data = {
    "Employee_ID": [101, 102, 103, 104, 105],

    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],

    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],

    "Age": [29, 34, 41, 28, 38],

    "Salary": [50000, 70000, 65000, 55000, 60000],

    "Years_of_Experience": [4, 8, 10, 3, 12],

    "Joining_Date": [
        "2020-03-15",
        "2017-07-19",
        "2013-06-01",
        "2021-02-10",
        "2010-11-25"
    ],

    "Gender": ["Female", "Male", "Male", "Female", "Male"],

    "Bonus": [5000, 7000, 6000, 4500, 5000],

    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5]
}

employees = pd.DataFrame(data)

print("Employee DataFrame:")
print(employees)




print("\na) Shape:")
print(employees.shape)




print("\nb) DataFrame Information:")
employees.info()




print("\nc) Descriptive Statistics:")
print(employees.describe())




print("\nd) First 5 rows:")
print(employees.head(5))

print("\nLast 3 rows:")
print(employees.tail(3))




average_salary = employees["Salary"].mean()

print("\ni) Average Salary:")
print(average_salary)




total_bonus = employees["Bonus"].sum()

print("\nii) Total Bonus:")
print(total_bonus)




youngest_age = employees["Age"].min()

print("\niii) Youngest Employee Age:")
print(youngest_age)




highest_rating = employees["Rating"].max()

print("\niv) Highest Performance Rating:")
print(highest_rating)




employees = employees.sort_values(
    by="Salary",
    ascending=False
)

print("\nf) Sorted by Salary (Descending):")
print(employees)




def performance_category(rating):

    if rating >= 4.5:
        return "Excellent"

    elif rating >= 4.0:
        return "Good"

    else:
        return "Average"


employees["Performance_Category"] = employees["Rating"].apply(
    performance_category
)

print("\ng) Performance Category:")
print(
    employees[
        ["Name", "Rating", "Performance_Category"]
    ]
)




print("\nh) Missing Values:")
print(employees.isnull().sum())




employees = employees.rename(
    columns={"Employee_ID": "ID"}
)

print("\ni) After Renaming Employee_ID:")
print(employees)



print("\nj-i) Employees with more than 5 years experience:")

experience_more_than_5 = employees[
    employees["Years_of_Experience"] > 5
]

print(experience_more_than_5)




print("\nj-ii) Employees in IT department:")

it_employees = employees[
    employees["Department"] == "IT"
]

print(it_employees)




employees["Tax"] = employees["Salary"] * 0.10

print("\nk) DataFrame after adding Tax:")
print(employees)




employees.to_csv(
    "employees_modified.csv",
    index=False
)

print("\nl) Modified DataFrame saved as employees_modified.csv")
```
