```python
import pandas as pd

fixed_entries = [
    {
        "question": "what is the annual fee",
        "answer": "The annual fee is Rs 500.",
        "keywords": "fee cost price charge",
        "category": "billing"
    },
    {
        "question": "how to reset password",
        "answer": "Go to Settings > Reset Password.",
        "keywords": "password reset login",
        "category": "account"
    },
    {
        "question": "what are your working hours",
        "answer": "We are open 9 AM to 5 PM.",
        "keywords": "hours timing open time",
        "category": "general"
    },
    {
        "question": "how can i pay the fee",
        "answer": "You can pay via UPI, card, or net banking.",
        "keywords": "pay payment upi fee",
        "category": "billing"
    }
]

personalized_entries = [
    {
        "question": "how can i check my billing details",
        "answer": "You can check your billing details from the billing section.",
        "keywords": "billing details invoice bill",
        "category": "billing"
    },
    {
        "question": "where can i find general information",
        "answer": "You can find general information in the help section.",
        "keywords": "information help details support",
        "category": "general"
    }
]

all_entries = fixed_entries + personalized_entries

df = pd.DataFrame(all_entries)

print(df)
```

Q2)

```python
def score_query(query, df):

    query_words = set(query.lower().split())

    results = []

    for index, row in df.iterrows():

        keywords = set(row["keywords"].lower().split())

        score = len(query_words & keywords)

        if score > 0:
            results.append({
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "score": score
            })

    results.sort(
        key=lambda x: x["score"],
        reverse=True
    )

    return results


query = input("Enter your question: ")

results = score_query(query, df)

for result in results:
    print(result)
```

```python
def score_query_with_ties(query, df):

    query_words = set(query.lower().split())

    results = []

    for index, row in df.iterrows():

        keywords = set(row["keywords"].lower().split())

        score = len(query_words & keywords)

        if score > 0:
            results.append({
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "score": score
            })

    if len(results) == 0:
        print("No matching entries.")
        return

    highest_score = max(
        result["score"] for result in results
    )

    best_matches = [
        result
        for result in results
        if result["score"] == highest_score
    ]

    print("Highest Score:", highest_score)

    if len(best_matches) > 1:
        print("Tie detected!")

    for result in best_matches:

        print("\nQuestion:", result["question"])
        print("Answer:", result["answer"])
        print("Category:", result["category"])
        print("Score:", result["score"])
```

```python
score_query_with_ties("fee", df)
```
