```python
import random
random_list =[]
for i in range(0,100):
   random_list.append(random.randint(100,900))
print(random_list)
```

```python
odd=[]
for x in random_list:
  if(x%2==1):
    odd.append(x)
print(odd)
print(len(odd))
```

```python
even=[]
for x in random_list:
  if(x%2==0):
    even.append(x)
print(even)
print(len(even))
```

```python
def is_prime(num):
    if num <= 1:
        return False
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            return False
    return True
prime_list=[x for x in random_list if is_prime(x)]
print(prime_list)
```

```python
max_count=0
y=0
for x in random_list :
  if(random_list.count(x)>max_count):
    max_count=random_list.count(x)
    y=x
print(max_count)
print(y)
```

```python
num = [1, 0, 2, 4, 1, 6, 0, 0,0,5]


A = set(map(lambda x: x * 7, num))
B = set(map(lambda x: x * 9, num))

print("Set A:", A)
print("Set B:", B)
```

```python
union = A.union(B)
print("Union:", union)
```

```python
intersection = A.intersection(B)
print( intersection)
```

```python
A_minus_B = A.difference(B)
B_minus_A = B.difference(A)

print( A_minus_B)
print(B_minus_A)
```

```python
symmetric = A.symmetric_difference(B)
print("Symmetric difference:", symmetric)
```

```python
print("Is A a subset of B?", A.issubset(B))
print("Is B a superset of A?", B.issuperset(A))
```

```python
X = int(input("Enter a value X: "))

A.discard(X)

print("Set A after discard:", A)
```

```python
my_dict = {
    "name": "Amish Kapoor",
    "roll_no": "1024160005",
    "branch": "CSE",
    "age": 21,
    "city": "Shimla"
}

print("Original dictionary:")
print(my_dict)
```

```python
my_dict["location"] = my_dict.pop("city")

print("\nAfter renaming city to location:")
print(my_dict)
```

```python
my_dict["cgpa"] = 10

print("\nAfter adding CGPA:")
print(my_dict)
```

```python
my_dict["age"] += 1

print("\nAfter increasing age:")
print(my_dict)
```

```python
dict_pop = my_dict.copy()
dict_del = my_dict.copy()
```

```python
removed_branch = dict_pop.pop("branch")
del dict_del["branch"]

print("\nUsing pop():")
print(dict_pop)
print("Removed value:", removed_branch)

print("\nUsing del:")
print(dict_del)
```

```python
print("\nKey-value pairs:")

for key, value in my_dict.items():
    print(key, "→", value)
```

```python
if "email" in my_dict:
    print("\nEmail:", my_dict["email"])
else:
    print("\nEmail not found.")
```

```python
friend_dict = {
    "name": "Rahul Sharma",
    "roll_no": "1024160010",
    "branch": "CSE",
    "age": 21,
    "city": "Delhi"
}
merged_dict = {**my_dict, **friend_dict}

print("\nMerged dictionary:")
print(merged_dict)
```

```python
string_values = {
    key: value
    for key, value in my_dict.items()
    if isinstance(value, str)
}

print("\nOnly string values:")
print(string_values)
```
